# Notebook 1B - ETL Watcher (Loop)

Objetivo:
- Executar o mesmo ETL do Notebook 1A em loop.
- Processar automaticamente arquivos novos na pasta `data/pending`.


## Observacao de arquitetura

- O `01A` mostra a logica didatica.
- O `01B` operacionaliza execucao continua com base em `src/etl_core.py`.
- Isso permite demonstrar evolucao de prototipo para automacao.


```python
import os
import time
from datetime import datetime
from etl_core import run_once

INPUT_DIR = os.getenv("INPUT_DIR", "/workspace/data/pending")
PROCESSED_DIR = os.getenv("PROCESSED_DIR", "/workspace/data/processed")
FAILED_DIR = os.getenv("FAILED_DIR", "/workspace/data/failed")
```

```python
def run_n_cycles(cycles=3, sleep_seconds=20):
    for i in range(cycles):
        print(f"\n=== ciclo {i+1}/{cycles} | {datetime.now()} ===")
        results = run_once(INPUT_DIR, PROCESSED_DIR, FAILED_DIR)
        if not results:
            print("Sem arquivos para processar")
        for r in results:
            print(f"{r.file_name:40} | {r.status:8} | linhas: {r.rows_loaded}")
        if i < cycles - 1:
            time.sleep(sleep_seconds)

run_n_cycles(cycles=2, sleep_seconds=10)
```
